## 9. Relaciones y conservación

Gold utiliza contratos; procesos, proveedores y grupos se conservan como bases auxiliares independientes. Los agregados C/R/A se cruzan con contratos por claves y segmentos, sin unir por nombres. Se comprueba unicidad de id_contrato y conservación del número de contratos tras los cruces. IRCD solo suma si los tres scores existen. La falta de causal o de comparabilidad no se convierte en cero.
